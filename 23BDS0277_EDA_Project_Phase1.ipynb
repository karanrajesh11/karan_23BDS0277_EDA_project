```python
# EDA Course Project - Phase 1
# Name : Karan
# Reg No : 23BDS0277

# importing required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# loading the dataset

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/USSeatBelts.csv"

dataset_karan = pd.read_csv(url)

# displaying dataset

print("First 5 Rows")
print(dataset_karan.head())

print()

print("Last 5 Rows")
print(dataset_karan.tail())

print()

print("Random Sample")
print(dataset_karan.sample(5, random_state=10))

print()

# dataset shape

print("Dataset Shape")
print("Rows :", dataset_karan.shape[0])
print("Columns :", dataset_karan.shape[1])

print()

# column names

print("Column Names")

print(dataset_karan.columns)

print()

# data types

print("Data Types")

print(dataset_karan.dtypes)

print()

# dataset information

print("Dataset Information")

dataset_karan.info()

print()

# basic statistics

print("Summary Statistics")

summary_statistics_23BDS0277 = dataset_karan.describe()

print(summary_statistics_23BDS0277)

print()

print("Summary Statistics (Including Categorical)")

print(dataset_karan.describe(include='all'))

print()

# mean

print("Mean")

print(dataset_karan.mean(numeric_only=True))

print()

# median

print("Median")

print(dataset_karan.median(numeric_only=True))

print()

# mode

print("Mode")

print(dataset_karan.mode().iloc[0])

print()

# variance

print("Variance")

print(dataset_karan.var(numeric_only=True))

print()

# standard deviation

print("Standard Deviation")

print(dataset_karan.std(numeric_only=True))

print()

# minimum values

print("Minimum Values")

print(dataset_karan.min(numeric_only=True))

print()

# maximum values

print("Maximum Values")

print(dataset_karan.max(numeric_only=True))

print()

# unique values

print("Unique Values")

for column in dataset_karan.columns:
    print(column, ":", dataset_karan[column].nunique())

print()

# value counts

print("State Value Counts")

print(dataset_karan["state"].value_counts())

print()

print("Enforcement Value Counts")

print(dataset_karan["enforce"].value_counts())

print()

# checking missing values

print("Missing Values")

missing_values_23BDS0277 = dataset_karan.isnull().sum()

print(missing_values_23BDS0277)

print()

missing_percentage_23BDS0277 = (dataset_karan.isnull().sum() / len(dataset_karan)) * 100

print("Missing Percentage")

print(missing_percentage_23BDS0277)

print()

# missing value heatmap

plt.figure(figsize=(10,5))

sns.heatmap(dataset_karan.isnull(),
            cbar=False,
            cmap="viridis")

plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Records")
plt.tight_layout()
plt.show()

print("Observation")
print("- Seatbelt column contains missing values.")
print("- Other columns have very few or no missing values.")

print()

# checking duplicates

print("Duplicate Records")

duplicate_rows_23BDS0277 = dataset_karan.duplicated().sum()

print("Duplicate Rows :", duplicate_rows_23BDS0277)

print()

# removing duplicates

cleaned_data_karan = dataset_karan.drop_duplicates()

print("Shape After Removing Duplicates")

print(cleaned_data_karan.shape)

print()

# removing unwanted column

if "rownames" in cleaned_data_karan.columns:
    cleaned_data_karan = cleaned_data_karan.drop(columns=["rownames"])

print("Columns After Removing rownames")

print(cleaned_data_karan.columns)

print()

# handling missing values

numeric_columns_23BDS0277 = cleaned_data_karan.select_dtypes(include=np.number).columns

for column in numeric_columns_23BDS0277:
    cleaned_data_karan[column] = cleaned_data_karan[column].fillna(cleaned_data_karan[column].median())

categorical_columns_23BDS0277 = cleaned_data_karan.select_dtypes(include="object").columns

for column in categorical_columns_23BDS0277:
    cleaned_data_karan[column] = cleaned_data_karan[column].fillna(cleaned_data_karan[column].mode()[0])

print("Missing Values After Cleaning")

print(cleaned_data_karan.isnull().sum())

print()

print("Observation")
print("- Missing values have been handled.")
print("- Numeric values were filled using median.")
print("- Categorical values were filled using mode.")
```

```python
# data transformation

print("Data Transformation")

transformed_data_karan = cleaned_data_karan.copy()

# checking categorical columns

print("Categorical Columns")

categorical_columns_23BDS0277 = transformed_data_karan.select_dtypes(include="object").columns

print(categorical_columns_23BDS0277)

print()

# label encoding

print("Label Encoding")

yes_no_columns = [
    "seatbelt",
    "speed65",
    "speed70",
    "drinkage",
    "alcohol"
]

for column in yes_no_columns:
    if column in transformed_data_karan.columns:
        transformed_data_karan[column] = transformed_data_karan[column].map({
            "yes": 1,
            "no": 0
        })

if "enforce" in transformed_data_karan.columns:
    transformed_data_karan["enforce_code"] = transformed_data_karan["enforce"].map({
        "no": 0,
        "secondary": 1,
        "primary": 2
    })

print("Label Encoding Completed")

print()

# one hot encoding

print("One Hot Encoding")

encoded_data_23BDS0277 = pd.get_dummies(
    transformed_data_karan,
    columns=["state"],
    drop_first=True
)

print("Shape Before Encoding :", transformed_data_karan.shape)
print("Shape After Encoding :", encoded_data_23BDS0277.shape)

print()

# creating new feature

print("Creating New Feature")

transformed_data_karan["fatality_rate"] = (
    transformed_data_karan["fatalities"] /
    transformed_data_karan["miles"]
) * 10000

print(
    transformed_data_karan[
        ["fatalities", "miles", "fatality_rate"]
    ].head()
)

print()

# creating income category

print("Creating Income Category")

transformed_data_karan["income_category"] = pd.qcut(
    transformed_data_karan["income"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)

print(
    transformed_data_karan[
        ["income", "income_category"]
    ].head()
)

print()

# creating age category

print("Creating Age Category")

transformed_data_karan["age_category"] = pd.qcut(
    transformed_data_karan["age"],
    q=4,
    labels=[
        "Young",
        "Middle Aged",
        "Senior",
        "Older"
    ]
)

print(
    transformed_data_karan[
        ["age", "age_category"]
    ].head()
)

print()

# standardization

print("Standardization")

standardized_data_23BDS0277 = transformed_data_karan.copy()

numeric_columns_23BDS0277 = standardized_data_23BDS0277.select_dtypes(include=np.number).columns

for column in numeric_columns_23BDS0277:

    if standardized_data_23BDS0277[column].std() != 0:

        standardized_data_23BDS0277[column] = (
            standardized_data_23BDS0277[column] -
            standardized_data_23BDS0277[column].mean()
        ) / standardized_data_23BDS0277[column].std()

print("Standardization Completed")

print()

# transformed dataset

print("Transformed Dataset")

print(transformed_data_karan.head())

print()

print("Shape After Transformation")

print(transformed_data_karan.shape)

print()

```

```python
# UNIVARIATE ANALYSIS

print("UNIVARIATE ANALYSIS")

# Histogram - Fatalities

plt.figure(figsize=(8,5))

sns.histplot(
    transformed_data_karan["fatalities"],
    bins=25,
    kde=True,
    color="steelblue"
)

plt.title("Distribution of Fatalities")
plt.xlabel("Fatalities")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("Observation")
print("- The histogram shows the distribution of fatality rates.")
print("- Most observations are concentrated at lower fatality values.")
print("- The distribution is slightly right skewed.")
print()

# Histogram - Income

plt.figure(figsize=(8,5))

sns.histplot(
    transformed_data_karan["income"],
    bins=25,
    kde=True,
    color="orange"
)

plt.title("Distribution of Income")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("Observation")
print("- Income values are spread across different ranges.")
print("- Most observations lie around the middle of the distribution.")
print("- A few higher income values are present.")
print()

# Box Plot - Miles

plt.figure(figsize=(8,5))

sns.boxplot(
    x=transformed_data_karan["miles"],
    color="lightgreen"
)

plt.title("Box Plot of Miles")
plt.xlabel("Miles")
plt.tight_layout()
plt.show()

print("Observation")
print("- The box plot shows the spread of miles travelled.")
print("- A few outliers are present.")
print("- Most observations fall within the interquartile range.")
print()

# Count Plot - Enforcement

plt.figure(figsize=(8,5))

sns.countplot(
    data=transformed_data_karan,
    x="enforce",
    palette="Set2"
)

plt.title("Distribution of Enforcement Type")
plt.xlabel("Enforcement")
plt.ylabel("Count")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

print("Observation")
print("- The count plot shows the frequency of each enforcement category.")
print("- The categories are not equally distributed.")
print("- One enforcement type occurs more frequently than the others.")
print()

# Pie Chart - Enforcement

enforce_counts = transformed_data_karan["enforce"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    enforce_counts,
    labels=enforce_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Enforcement Distribution")
plt.tight_layout()
plt.show()

print("Observation")
print("- The pie chart shows the percentage share of each enforcement category.")
print("- One enforcement category contributes the highest proportion.")
print("- The remaining categories make up the rest of the dataset.")
print()

print("Univariate Analysis Completed")
```

```python
# BIVARIATE ANALYSIS

print("BIVARIATE ANALYSIS")

# Scatter Plot - Income vs Fatalities

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=transformed_data_karan,
    x="income",
    y="fatalities",
    color="steelblue"
)

plt.title("Income vs Fatalities")
plt.xlabel("Income")
plt.ylabel("Fatalities")
plt.tight_layout()
plt.show()

print("Observation")
print("- The scatter plot shows the relationship between income and fatalities.")
print("- The points are widely scattered, indicating a weak relationship.")
print("- No strong linear trend can be observed.")
print()

# Scatter Plot - Miles vs Fatalities

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=transformed_data_karan,
    x="miles",
    y="fatalities",
    color="darkorange"
)

plt.title("Miles vs Fatalities")
plt.xlabel("Miles")
plt.ylabel("Fatalities")
plt.tight_layout()
plt.show()

print("Observation")
print("- Fatalities generally increase with the number of miles travelled.")
print("- Some observations lie far from the main cluster.")
print("- A few possible outliers are present.")
print()

# Regression Plot

plt.figure(figsize=(8,5))

sns.regplot(
    data=transformed_data_karan,
    x="income",
    y="fatalities",
    scatter_kws={"alpha":0.6},
    line_kws={"color":"red"}
)

plt.title("Regression Plot of Income and Fatalities")
plt.xlabel("Income")
plt.ylabel("Fatalities")
plt.tight_layout()
plt.show()

print("Observation")
print("- The regression line shows the overall relationship between the variables.")
print("- The slope indicates a moderately strong negative correlation between the variables as there are certain concentrated regions on the slope and certain sparse regions.")
print("- Most observations are spread around the regression line.")
print()

# Box Plot - Enforcement vs Fatalities

plt.figure(figsize=(8,5))

sns.boxplot(
    data=transformed_data_karan,
    x="enforce",
    y="fatalities",
    palette="Set3"
)

plt.title("Fatalities by Enforcement Type")
plt.xlabel("Enforcement")
plt.ylabel("Fatalities")
plt.tight_layout()
plt.show()

print("Observation")
print("- Fatality distribution varies across different enforcement categories.")
print("- Median values differ slightly among the groups.")
print("- Outliers are present in some enforcement categories.")
print()

# Line Plot - Median Fatalities by Year

median_year_karan = (
    transformed_data_karan
    .groupby("year")["fatalities"]
    .median()
    .reset_index()
)

plt.figure(figsize=(10,5))

sns.lineplot(
    data=median_year_karan,
    x="year",
    y="fatalities",
    marker="o"
)

plt.title("Median Fatalities by Year")
plt.xlabel("Year")
plt.ylabel("Median Fatalities")
plt.tight_layout()
plt.show()

print("Observation")
print("- The line plot shows how median fatalities change over time.")
print("- There are noticeable fluctuations across different years.")
print("- Overall trends can be identified from the graph.")
print()

print()
```

```python
# MULTIVARIATE ANALYSIS

print("MULTIVARIATE ANALYSIS")

# Correlation Heatmap

numeric_data_23BDS0277 = transformed_data_karan.select_dtypes(include=np.number)

correlation_matrix_karan = numeric_data_23BDS0277.corr()

plt.figure(figsize=(10,8))

sns.heatmap(
    correlation_matrix_karan,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.xlabel("Features")
plt.ylabel("Features")
plt.tight_layout()
plt.show()

print("Observation")
print("- Positive and negative correlations can be identified.")
print("- Values close to 1 indicate strong positive correlation.")
print("- Values close to -1 indicate strong negative correlation.")
print("- Most variables have weak to moderate correlation.")
print()

# Pair Plot

pairplot_columns = [
    "fatalities",
    "miles",
    "income",
    "age"
]

sns.pairplot(
    transformed_data_karan[pairplot_columns],
    diag_kind="hist"
)

plt.show()

print("Observation")
print("- Pairwise relationships between numerical variables are visible.")
print("- Histograms on the diagonal show the distribution of each variable.")
print("- Scatter plots help identify trends and possible outliers.")
print("- Most variables show weak linear relationship.")
print()

# Pair Plot with Enforcement

pairplot_data_karan = transformed_data_karan[
    ["fatalities", "income", "miles", "enforce"]
]

sns.pairplot(
    pairplot_data_karan,
    hue="enforce",
    diag_kind="hist"
)

plt.show()

print("Observation")
print("- Data points are grouped based on enforcement type.")
print("- Different enforcement categories can be compared visually.")
print("- Considerable overlap exists between the enforcement groups.")
print("- The plot helps understand relationships among multiple variables.")
print()
```
