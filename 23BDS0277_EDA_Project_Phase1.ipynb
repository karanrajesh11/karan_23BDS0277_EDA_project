```python
# EDA Course Project - Phase 1
# Name : Karan
# Reg No : 23BDS0277

# importing required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# loading the dataset

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/USSeatBelts.csv"

dataset_karan = pd.read_csv(url)

# displaying dataset

print("First 5 Rows")
print(dataset_karan.head())

print()

print("Last 5 Rows")
print(dataset_karan.tail())

print()

print("Random Sample")
print(dataset_karan.sample(5, random_state=10))

print()

# dataset shape

print("Dataset Shape")
print("Rows :", dataset_karan.shape[0])
print("Columns :", dataset_karan.shape[1])

print()

# column names

print("Column Names")

print(dataset_karan.columns)

print()

# data types

print("Data Types")

print(dataset_karan.dtypes)

print()

# dataset information

print("Dataset Information")

dataset_karan.info()

print()

# basic statistics

print("Summary Statistics")

summary_statistics_23BDS0277 = dataset_karan.describe()

print(summary_statistics_23BDS0277)

print()

print("Summary Statistics (Including Categorical)")

print(dataset_karan.describe(include='all'))

print()

# mean

print("Mean")

print(dataset_karan.mean(numeric_only=True))

print()

# median

print("Median")

print(dataset_karan.median(numeric_only=True))

print()

# mode

print("Mode")

print(dataset_karan.mode().iloc[0])

print()

# variance

print("Variance")

print(dataset_karan.var(numeric_only=True))

print()

# standard deviation

print("Standard Deviation")

print(dataset_karan.std(numeric_only=True))

print()

# minimum values

print("Minimum Values")

print(dataset_karan.min(numeric_only=True))

print()

# maximum values

print("Maximum Values")

print(dataset_karan.max(numeric_only=True))

print()

# unique values

print("Unique Values")

for column in dataset_karan.columns:
    print(column, ":", dataset_karan[column].nunique())

print()

# value counts

print("State Value Counts")

print(dataset_karan["state"].value_counts())

print()

print("Enforcement Value Counts")

print(dataset_karan["enforce"].value_counts())

print()

# checking missing values

print("Missing Values")

missing_values_23BDS0277 = dataset_karan.isnull().sum()

print(missing_values_23BDS0277)

print()

missing_percentage_23BDS0277 = (dataset_karan.isnull().sum() / len(dataset_karan)) * 100

print("Missing Percentage")

print(missing_percentage_23BDS0277)

print()

# missing value heatmap

plt.figure(figsize=(10,5))

sns.heatmap(dataset_karan.isnull(),
            cbar=False,
            cmap="viridis")

plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Records")
plt.tight_layout()
plt.show()

print("Observation")
print("- Seatbelt column contains missing values.")
print("- Other columns have very few or no missing values.")

print()

# checking duplicates

print("Duplicate Records")

duplicate_rows_23BDS0277 = dataset_karan.duplicated().sum()

print("Duplicate Rows :", duplicate_rows_23BDS0277)

print()

# removing duplicates

cleaned_data_karan = dataset_karan.drop_duplicates()

print("Shape After Removing Duplicates")

print(cleaned_data_karan.shape)

print()

# removing unwanted column

if "rownames" in cleaned_data_karan.columns:
    cleaned_data_karan = cleaned_data_karan.drop(columns=["rownames"])

print("Columns After Removing rownames")

print(cleaned_data_karan.columns)

print()

# handling missing values

numeric_columns_23BDS0277 = cleaned_data_karan.select_dtypes(include=np.number).columns

for column in numeric_columns_23BDS0277:
    cleaned_data_karan[column] = cleaned_data_karan[column].fillna(cleaned_data_karan[column].median())

categorical_columns_23BDS0277 = cleaned_data_karan.select_dtypes(include="object").columns

for column in categorical_columns_23BDS0277:
    cleaned_data_karan[column] = cleaned_data_karan[column].fillna(cleaned_data_karan[column].mode()[0])

print("Missing Values After Cleaning")

print(cleaned_data_karan.isnull().sum())

print()

print("Observation")
print("- Missing values have been handled.")
print("- Numeric values were filled using median.")
print("- Categorical values were filled using mode.")
```

```python
# data transformation

print("Data Transformation")

transformed_data_karan = cleaned_data_karan.copy()

# checking categorical columns

print("Categorical Columns")

categorical_columns_23BDS0277 = transformed_data_karan.select_dtypes(include="object").columns

print(categorical_columns_23BDS0277)

print()

# label encoding

print("Label Encoding")

yes_no_columns = [
    "seatbelt",
    "speed65",
    "speed70",
    "drinkage",
    "alcohol"
]

for column in yes_no_columns:
    if column in transformed_data_karan.columns:
        transformed_data_karan[column] = transformed_data_karan[column].map({
            "yes": 1,
            "no": 0
        })

if "enforce" in transformed_data_karan.columns:
    transformed_data_karan["enforce_code"] = transformed_data_karan["enforce"].map({
        "no": 0,
        "secondary": 1,
        "primary": 2
    })

print("Label Encoding Completed")

print()

# one hot encoding

print("One Hot Encoding")

encoded_data_23BDS0277 = pd.get_dummies(
    transformed_data_karan,
    columns=["state"],
    drop_first=True
)

print("Shape Before Encoding :", transformed_data_karan.shape)
print("Shape After Encoding :", encoded_data_23BDS0277.shape)

print()

# creating new feature

print("Creating New Feature")

transformed_data_karan["fatality_rate"] = (
    transformed_data_karan["fatalities"] /
    transformed_data_karan["miles"]
) * 10000

print(
    transformed_data_karan[
        ["fatalities", "miles", "fatality_rate"]
    ].head()
)

print()

# creating income category

print("Creating Income Category")

transformed_data_karan["income_category"] = pd.qcut(
    transformed_data_karan["income"],
    q=4,
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)

print(
    transformed_data_karan[
        ["income", "income_category"]
    ].head()
)

print()

# creating age category

print("Creating Age Category")

transformed_data_karan["age_category"] = pd.qcut(
    transformed_data_karan["age"],
    q=4,
    labels=[
        "Young",
        "Middle Aged",
        "Senior",
        "Older"
    ]
)

print(
    transformed_data_karan[
        ["age", "age_category"]
    ].head()
)

print()

# standardization

print("Standardization")

standardized_data_23BDS0277 = transformed_data_karan.copy()

numeric_columns_23BDS0277 = standardized_data_23BDS0277.select_dtypes(include=np.number).columns

for column in numeric_columns_23BDS0277:

    if standardized_data_23BDS0277[column].std() != 0:

        standardized_data_23BDS0277[column] = (
            standardized_data_23BDS0277[column] -
            standardized_data_23BDS0277[column].mean()
        ) / standardized_data_23BDS0277[column].std()

print("Standardization Completed")

print()

# transformed dataset

print("Transformed Dataset")

print(transformed_data_karan.head())

print()

print("Shape After Transformation")

print(transformed_data_karan.shape)

print()

```

```python
# EDA Course Project - Phase 2
# Name : Karan Rajesh
# Reg No : 23BDS0277


# importing required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist, squareform
from scipy.sparse.csgraph import minimum_spanning_tree

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering, SpectralClustering, BisectingKMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, calinski_harabasz_score
from sklearn.ensemble import IsolationForest

import warnings
warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")


# loading the dataset

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/USSeatBelts.csv"

dataset_karan = pd.read_csv(url)




# preparing data for Phase 2

phase2_data_karan = dataset_karan.copy()

# removing unwanted row number column

if "rownames" in phase2_data_karan.columns:
    phase2_data_karan = phase2_data_karan.drop(columns=["rownames"])


# handling missing values

numeric_columns_23BDS0277 = phase2_data_karan.select_dtypes(include=np.number).columns

for column in numeric_columns_23BDS0277:
    phase2_data_karan[column] = phase2_data_karan[column].fillna(
        phase2_data_karan[column].median()
    )


categorical_columns_23BDS0277 = phase2_data_karan.select_dtypes(include="object").columns

for column in categorical_columns_23BDS0277:
    phase2_data_karan[column] = phase2_data_karan[column].fillna(
        phase2_data_karan[column].mode()[0]
    )


# converting yes and no variables into numerical form

binary_columns_23BDS0277 = [
    "speed65",
    "speed70",
    "drinkage",
    "alcohol"
]

for column in binary_columns_23BDS0277:

    if column in phase2_data_karan.columns:

        phase2_data_karan[column + "_code"] = phase2_data_karan[column].map({
            "yes": 1,
            "no": 0
        })


# converting enforcement into numerical form

phase2_data_karan["enforce_code"] = phase2_data_karan["enforce"].map({
    "no": 0,
    "secondary": 1,
    "primary": 2
})


# creating fatality rate

phase2_data_karan["fatality_rate"] = (
    phase2_data_karan["fatalities"] /
    phase2_data_karan["miles"]
) * 10000










# selecting important numerical variables

numeric_analysis_columns = [
    "year",
    "miles",
    "fatalities",
    "seatbelt",
    "income",
    "age",
    "fatality_rate"
]

numeric_analysis_columns = [
    column for column in numeric_analysis_columns
    if column in phase2_data_karan.columns
]


# ------------------------------------------------------------
# Measures of central tendency and dispersion
# ------------------------------------------------------------


statistics_table_karan = pd.DataFrame({
    "Mean": phase2_data_karan[numeric_analysis_columns].mean(),
    "Median": phase2_data_karan[numeric_analysis_columns].median(),
    "Minimum": phase2_data_karan[numeric_analysis_columns].min(),
    "Maximum": phase2_data_karan[numeric_analysis_columns].max(),
    "Range": (
        phase2_data_karan[numeric_analysis_columns].max()
        -
        phase2_data_karan[numeric_analysis_columns].min()
    ),
    "Variance": phase2_data_karan[numeric_analysis_columns].var(),
    "Standard Deviation": phase2_data_karan[numeric_analysis_columns].std(),
    "IQR": (
        phase2_data_karan[numeric_analysis_columns].quantile(0.75)
        -
        phase2_data_karan[numeric_analysis_columns].quantile(0.25)
    )
})




# ------------------------------------------------------------
# Coefficient of variation
# ------------------------------------------------------------


coefficient_variation_karan = (
    phase2_data_karan[numeric_analysis_columns].std()
    /
    phase2_data_karan[numeric_analysis_columns].mean()
) * 100




# ------------------------------------------------------------
# Skewness and kurtosis
# ------------------------------------------------------------









# ------------------------------------------------------------
# Complete statistical summary
# ------------------------------------------------------------

shape_statistics_karan = pd.DataFrame({
    "Skewness": phase2_data_karan[numeric_analysis_columns].skew(),
    "Kurtosis": phase2_data_karan[numeric_analysis_columns].kurt(),
    "IQR": (
        phase2_data_karan[numeric_analysis_columns].quantile(0.75)
        -
        phase2_data_karan[numeric_analysis_columns].quantile(0.25)
    ),
    "CV (%)": coefficient_variation_karan
})




# ------------------------------------------------------------
# Frequency distribution
# ------------------------------------------------------------


fatality_frequency_karan = pd.cut(
    phase2_data_karan["fatalities"],
    bins=10
).value_counts().sort_index()




# ------------------------------------------------------------
# Frequency polygon
# ------------------------------------------------------------

frequency_counts_karan, frequency_bins_karan = np.histogram(
    phase2_data_karan["fatalities"],
    bins=10
)

frequency_midpoints_karan = (
    frequency_bins_karan[:-1]
    +
    frequency_bins_karan[1:]
) / 2

plt.figure(figsize=(9,5))

plt.plot(
    frequency_midpoints_karan,
    frequency_counts_karan,
    marker="o"
)

plt.title("Frequency Polygon of Fatalities")
plt.xlabel("Fatalities")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Additional boxplots for all important numerical variables
# ------------------------------------------------------------

plt.figure(figsize=(12,6))

sns.boxplot(
    data=phase2_data_karan[numeric_analysis_columns]
)

plt.title("Boxplots of Numerical Variables")
plt.xlabel("Variables")
plt.ylabel("Values")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Violin plot - numerical distributions
# ------------------------------------------------------------

violin_data_karan = phase2_data_karan[numeric_analysis_columns].melt(
    var_name="Variable",
    value_name="Value"
)

plt.figure(figsize=(12,6))

sns.violinplot(
    data=violin_data_karan,
    x="Variable",
    y="Value"
)

plt.title("Violin Plots of Numerical Variables")
plt.xlabel("Variable")
plt.ylabel("Value")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# ECDF - Empirical Cumulative Distribution Function
# ------------------------------------------------------------

plt.figure(figsize=(9,5))

sns.ecdfplot(
    data=phase2_data_karan,
    x="fatalities"
)

plt.title("ECDF of Fatalities")
plt.xlabel("Fatalities")
plt.ylabel("Cumulative Probability")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# ECDF - Income
# ------------------------------------------------------------

plt.figure(figsize=(9,5))

sns.ecdfplot(
    data=phase2_data_karan,
    x="income"
)

plt.title("ECDF of Income")
plt.xlabel("Income")
plt.ylabel("Cumulative Probability")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# KDE plot - Age
# ------------------------------------------------------------

plt.figure(figsize=(9,5))

sns.kdeplot(
    data=phase2_data_karan,
    x="age",
    fill=True
)

plt.title("Kernel Density Plot of Age")
plt.xlabel("Age")
plt.ylabel("Density")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# KDE plot - Seatbelt usage
# ------------------------------------------------------------

plt.figure(figsize=(9,5))

sns.kdeplot(
    data=phase2_data_karan,
    x="seatbelt",
    fill=True
)

plt.title("Kernel Density Plot of Seatbelt Usage")
plt.xlabel("Seatbelt Usage")
plt.ylabel("Density")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Q-Q plot - Fatalities
# ------------------------------------------------------------

plt.figure(figsize=(7,7))

stats.probplot(
    phase2_data_karan["fatalities"],
    dist="norm",
    plot=plt
)

plt.title("Q-Q Plot of Fatalities")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Q-Q plot - Income
# ------------------------------------------------------------

plt.figure(figsize=(7,7))

stats.probplot(
    phase2_data_karan["income"],
    dist="norm",
    plot=plt
)

plt.title("Q-Q Plot of Income")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# State frequency distribution
# ------------------------------------------------------------

state_frequency_karan = (
    phase2_data_karan["state"]
    .value_counts()
    .sort_values(ascending=True)
)

plt.figure(figsize=(10,12))

state_frequency_karan.plot(kind="barh")

plt.title("Frequency Distribution of States")
plt.xlabel("Frequency")
plt.ylabel("State")
plt.tight_layout()
plt.show()






# yearly statistical characteristics

yearly_statistics_karan = (
    phase2_data_karan
    .groupby("year")["fatalities"]
    .agg([
        "mean",
        "median",
        "min",
        "max",
        "var",
        "std"
    ])
)

yearly_statistics_karan["range"] = (
    yearly_statistics_karan["max"]
    -
    yearly_statistics_karan["min"]
)




# ------------------------------------------------------------
# Mean and median fatalities over time
# ------------------------------------------------------------

plt.figure(figsize=(11,5))

plt.plot(
    yearly_statistics_karan.index,
    yearly_statistics_karan["mean"],
    marker="o",
    label="Mean"
)

plt.plot(
    yearly_statistics_karan.index,
    yearly_statistics_karan["median"],
    marker="s",
    label="Median"
)

plt.title("Mean and Median Fatalities by Year")
plt.xlabel("Year")
plt.ylabel("Fatalities")
plt.legend()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Yearly variance and standard deviation
# ------------------------------------------------------------

plt.figure(figsize=(11,5))

plt.plot(
    yearly_statistics_karan.index,
    yearly_statistics_karan["var"],
    marker="o",
    label="Variance"
)

plt.plot(
    yearly_statistics_karan.index,
    yearly_statistics_karan["std"],
    marker="s",
    label="Standard Deviation"
)

plt.title("Variance and Standard Deviation of Fatalities by Year")
plt.xlabel("Year")
plt.ylabel("Value")
plt.legend()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Yearly range
# ------------------------------------------------------------

plt.figure(figsize=(11,5))

plt.plot(
    yearly_statistics_karan.index,
    yearly_statistics_karan["range"],
    marker="o"
)

plt.title("Range of Fatalities by Year")
plt.xlabel("Year")
plt.ylabel("Range")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Sun ray style plot for yearly mean fatalities
# ------------------------------------------------------------

years_karan = yearly_statistics_karan.index.to_numpy()

mean_fatalities_karan = yearly_statistics_karan["mean"].to_numpy()

angles_karan = np.linspace(
    0,
    2 * np.pi,
    len(years_karan),
    endpoint=False
)

angles_karan = np.append(
    angles_karan,
    angles_karan[0]
)

values_karan = np.append(
    mean_fatalities_karan,
    mean_fatalities_karan[0]
)

plt.figure(figsize=(8,8))

ax = plt.subplot(111, polar=True)

ax.plot(
    angles_karan,
    values_karan,
    marker="o"
)

ax.fill(
    angles_karan,
    values_karan,
    alpha=0.15
)

ax.set_xticks(angles_karan[:-1])
ax.set_xticklabels(years_karan)

plt.title("Sun Ray Plot of Mean Fatalities by Year")

plt.show()






# ------------------------------------------------------------
# Additional scatter plots not present in Phase 1
# ------------------------------------------------------------

new_scatter_pairs_karan = [
    ("age", "income"),
    ("age", "fatalities"),
    ("age", "miles"),
    ("income", "miles"),
    ("seatbelt", "fatalities"),
    ("seatbelt", "income"),
    ("seatbelt", "miles"),
    ("seatbelt", "age")
]

for x_column, y_column in new_scatter_pairs_karan:

    plt.figure(figsize=(8,5))

    sns.scatterplot(
        data=phase2_data_karan,
        x=x_column,
        y=y_column
    )

    plt.title(
        x_column.capitalize()
        +
        " vs "
        +
        y_column.capitalize()
    )

    plt.xlabel(x_column.capitalize())
    plt.ylabel(y_column.capitalize())

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# Regression relationship - Age and Income
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.regplot(
    data=phase2_data_karan,
    x="age",
    y="income",
    scatter_kws={"alpha": 0.5}
)

plt.title("Regression Plot of Age and Income")
plt.xlabel("Age")
plt.ylabel("Income")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Hexbin plot
# ------------------------------------------------------------

plt.figure(figsize=(9,6))

plt.hexbin(
    phase2_data_karan["miles"],
    phase2_data_karan["income"],
    gridsize=25
)

plt.colorbar(label="Number of Observations")

plt.title("Hexbin Plot of Miles and Income")
plt.xlabel("Miles")
plt.ylabel("Income")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 2-D density plot
# ------------------------------------------------------------

plt.figure(figsize=(9,6))

sns.kdeplot(
    data=phase2_data_karan,
    x="age",
    y="income",
    fill=True
)

plt.title("2-D Density Plot of Age and Income")
plt.xlabel("Age")
plt.ylabel("Income")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Spearman correlation
# ------------------------------------------------------------

spearman_correlation_karan = (
    phase2_data_karan[numeric_analysis_columns]
    .corr(method="spearman")
)

plt.figure(figsize=(10,8))

sns.heatmap(
    spearman_correlation_karan,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Spearman Rank Correlation Heatmap")
plt.tight_layout()
plt.show()






# ------------------------------------------------------------
# Fatalities by speed limit 65 law
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.violinplot(
    data=phase2_data_karan,
    x="speed65",
    y="fatalities"
)

plt.title("Fatalities by 65 MPH Speed Limit")
plt.xlabel("65 MPH Speed Limit")
plt.ylabel("Fatalities")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Fatalities by 70 MPH speed limit
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.violinplot(
    data=phase2_data_karan,
    x="speed70",
    y="fatalities"
)

plt.title("Fatalities by 70 MPH Speed Limit")
plt.xlabel("70 MPH Speed Limit")
plt.ylabel("Fatalities")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Income by enforcement type
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    data=phase2_data_karan,
    x="enforce",
    y="income"
)

plt.title("Income Distribution by Enforcement Type")
plt.xlabel("Enforcement")
plt.ylabel("Income")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Age by enforcement type
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.violinplot(
    data=phase2_data_karan,
    x="enforce",
    y="age"
)

plt.title("Age Distribution by Enforcement Type")
plt.xlabel("Enforcement")
plt.ylabel("Age")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Strip plot
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.stripplot(
    data=phase2_data_karan,
    x="enforce",
    y="fatalities",
    jitter=True,
    alpha=0.5
)

plt.title("Fatalities by Enforcement Type")
plt.xlabel("Enforcement")
plt.ylabel("Fatalities")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Mean fatality rate by categorical variables
# ------------------------------------------------------------

categorical_analysis_columns = [
    "speed65",
    "speed70",
    "drinkage",
    "alcohol",
    "enforce"
]

for column in categorical_analysis_columns:

    grouped_fatalities_karan = (
        phase2_data_karan
        .groupby(column)["fatalities"]
        .mean()
        .sort_values()
    )

    plt.figure(figsize=(8,5))

    grouped_fatalities_karan.plot(
        kind="bar"
    )

    plt.title(
        "Mean Fatalities by "
        +
        column
    )

    plt.xlabel(column)
    plt.ylabel("Mean Fatalities")

    plt.xticks(rotation=20)

    plt.tight_layout()
    plt.show()






# ------------------------------------------------------------
# Cross tabulation
# ------------------------------------------------------------

cross_table_karan = pd.crosstab(
    phase2_data_karan["speed65"],
    phase2_data_karan["speed70"]
)




# ------------------------------------------------------------
# Row profiles
# ------------------------------------------------------------

row_profiles_karan = cross_table_karan.div(
    cross_table_karan.sum(axis=1),
    axis=0
)




# ------------------------------------------------------------
# Column profiles
# ------------------------------------------------------------

column_profiles_karan = cross_table_karan.div(
    cross_table_karan.sum(axis=0),
    axis=1
)




# ------------------------------------------------------------
# Contingency table heatmap
# ------------------------------------------------------------

plt.figure(figsize=(7,5))

sns.heatmap(
    cross_table_karan,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Contingency Table - Speed 65 and Speed 70")
plt.xlabel("Speed 70")
plt.ylabel("Speed 65")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Row profile heatmap
# ------------------------------------------------------------

plt.figure(figsize=(7,5))

sns.heatmap(
    row_profiles_karan,
    annot=True,
    fmt=".2f",
    cmap="Greens"
)

plt.title("Row Profiles - Speed 65 and Speed 70")
plt.xlabel("Speed 70")
plt.ylabel("Speed 65")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Stacked bar plot
# ------------------------------------------------------------

cross_table_karan.plot(
    kind="bar",
    stacked=True,
    figsize=(8,5)
)

plt.title("Stacked Bar Plot - Speed 65 and Speed 70")
plt.xlabel("Speed 65")
plt.ylabel("Frequency")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Chi-Square test
# ------------------------------------------------------------

chi_square_karan, p_value_karan, degrees_freedom_karan, expected_karan = (
    stats.chi2_contingency(cross_table_karan)
)




# ------------------------------------------------------------
# Expected frequencies
# ------------------------------------------------------------

expected_table_karan = pd.DataFrame(
    expected_karan,
    index=cross_table_karan.index,
    columns=cross_table_karan.columns
)




# ------------------------------------------------------------
# Cramer's V
# ------------------------------------------------------------

n_karan = cross_table_karan.to_numpy().sum()

cramers_v_karan = np.sqrt(
    chi_square_karan
    /
    (
        n_karan
        *
        min(
            cross_table_karan.shape[0] - 1,
            cross_table_karan.shape[1] - 1
        )
    )
)






# ------------------------------------------------------------
# 3-D scatter plot
# ------------------------------------------------------------

fig = plt.figure(figsize=(10,7))

ax = fig.add_subplot(111, projection="3d")

scatter_3d_karan = ax.scatter(
    phase2_data_karan["miles"],
    phase2_data_karan["income"],
    phase2_data_karan["fatalities"],
    c=phase2_data_karan["age"],
    cmap="viridis",
    s=40
)

ax.set_xlabel("Miles")
ax.set_ylabel("Income")
ax.set_zlabel("Fatalities")

plt.title("3-D Scatter Plot of Miles, Income and Fatalities")

fig.colorbar(
    scatter_3d_karan,
    ax=ax,
    label="Age"
)

plt.show()


# ------------------------------------------------------------
# 3-D scatter using seatbelt usage
# ------------------------------------------------------------

fig = plt.figure(figsize=(10,7))

ax = fig.add_subplot(111, projection="3d")

scatter_3d_karan = ax.scatter(
    phase2_data_karan["age"],
    phase2_data_karan["seatbelt"],
    phase2_data_karan["fatalities"],
    c=phase2_data_karan["income"],
    cmap="plasma",
    s=40
)

ax.set_xlabel("Age")
ax.set_ylabel("Seatbelt Usage")
ax.set_zlabel("Fatalities")

plt.title("3-D Scatter Plot of Age, Seatbelt Usage and Fatalities")

fig.colorbar(
    scatter_3d_karan,
    ax=ax,
    label="Income"
)

plt.show()


# ------------------------------------------------------------
# 3-D surface style plot
# ------------------------------------------------------------

surface_data_karan = phase2_data_karan[
    ["year", "miles", "fatalities"]
].sort_values("year")

fig = plt.figure(figsize=(10,7))

ax = fig.add_subplot(111, projection="3d")

ax.plot_trisurf(
    surface_data_karan["year"],
    surface_data_karan["miles"],
    surface_data_karan["fatalities"],
    cmap="viridis",
    alpha=0.8
)

ax.set_xlabel("Year")
ax.set_ylabel("Miles")
ax.set_zlabel("Fatalities")

plt.title("3-D Surface Representation of Year, Miles and Fatalities")

plt.show()






# selecting numerical features

nd_features_karan = [
    "miles",
    "fatalities",
    "seatbelt",
    "income",
    "age",
    "speed65_code",
    "speed70_code",
    "drinkage_code",
    "alcohol_code",
    "enforce_code"
]

nd_features_karan = [
    column
    for column in nd_features_karan
    if column in phase2_data_karan.columns
]


nd_data_karan = phase2_data_karan[
    nd_features_karan
].copy()


# standardization

scaler_karan = StandardScaler()

nd_scaled_karan = scaler_karan.fit_transform(
    nd_data_karan
)


# ------------------------------------------------------------
# PCA
# ------------------------------------------------------------

pca_karan = PCA()

pca_result_karan = pca_karan.fit_transform(
    nd_scaled_karan
)

explained_variance_karan = pca_karan.explained_variance_ratio_




# ------------------------------------------------------------
# PCA explained variance plot
# ------------------------------------------------------------

plt.figure(figsize=(9,5))

plt.bar(
    range(
        1,
        len(explained_variance_karan) + 1
    ),
    explained_variance_karan
)

plt.plot(
    range(
        1,
        len(explained_variance_karan) + 1
    ),
    np.cumsum(explained_variance_karan),
    marker="o"
)

plt.title("PCA Explained Variance")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# PCA 2-D projection
# ------------------------------------------------------------

pca_2d_karan = PCA(
    n_components=2
)

pca_2d_result_karan = pca_2d_karan.fit_transform(
    nd_scaled_karan
)

plt.figure(figsize=(9,6))

plt.scatter(
    pca_2d_result_karan[:, 0],
    pca_2d_result_karan[:, 1],
    alpha=0.6
)

plt.title("PCA 2-D Projection")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# PCA 3-D projection
# ------------------------------------------------------------

pca_3d_karan = PCA(
    n_components=3
)

pca_3d_result_karan = pca_3d_karan.fit_transform(
    nd_scaled_karan
)

fig = plt.figure(figsize=(10,7))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    pca_3d_result_karan[:, 0],
    pca_3d_result_karan[:, 1],
    pca_3d_result_karan[:, 2],
    alpha=0.6
)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

plt.title("PCA 3-D Projection")

plt.show()


# ------------------------------------------------------------
# PCA loading matrix
# ------------------------------------------------------------

pca_loading_karan = pd.DataFrame(
    pca_2d_karan.components_.T,
    index=nd_features_karan,
    columns=["PC1", "PC2"]
)




# ------------------------------------------------------------
# PCA biplot
# ------------------------------------------------------------

plt.figure(figsize=(10,7))

plt.scatter(
    pca_2d_result_karan[:, 0],
    pca_2d_result_karan[:, 1],
    alpha=0.35
)

for feature in nd_features_karan:

    loading_x = pca_loading_karan.loc[feature, "PC1"]
    loading_y = pca_loading_karan.loc[feature, "PC2"]

    plt.arrow(
        0,
        0,
        loading_x * 3,
        loading_y * 3,
        head_width=0.05,
        length_includes_head=True
    )

    plt.text(
        loading_x * 3.2,
        loading_y * 3.2,
        feature
    )

plt.title("PCA Biplot")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.axhline(0)
plt.axvline(0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Parallel coordinates plot
# ------------------------------------------------------------

parallel_data_karan = phase2_data_karan[
    [
        "miles",
        "fatalities",
        "seatbelt",
        "income",
        "age",
        "enforce"
    ]
].copy()

parallel_numeric_karan = [
    "miles",
    "fatalities",
    "seatbelt",
    "income",
    "age"
]

parallel_data_karan[parallel_numeric_karan] = StandardScaler().fit_transform(
    parallel_data_karan[parallel_numeric_karan]
)

parallel_data_karan = parallel_data_karan.sample(
    100,
    random_state=42
)

from pandas.plotting import parallel_coordinates

plt.figure(figsize=(12,7))

parallel_coordinates(
    parallel_data_karan,
    "enforce",
    alpha=0.35
)

plt.title("Parallel Coordinates Plot")
plt.xlabel("Variables")
plt.ylabel("Standardized Value")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()






# selecting exactly 50 instances

sample_50_karan = phase2_data_karan.sample(
    n=50,
    random_state=42
).copy()


# features used for clustering

clustering_features_karan = [
    "miles",
    "fatalities",
    "seatbelt",
    "income",
    "age",
    "speed65_code",
    "speed70_code",
    "drinkage_code",
    "alcohol_code",
    "enforce_code"
]

clustering_features_karan = [
    column
    for column in clustering_features_karan
    if column in sample_50_karan.columns
]


cluster_data_karan = sample_50_karan[
    clustering_features_karan
].copy()


# standardization

cluster_scaler_karan = StandardScaler()

cluster_scaled_karan = cluster_scaler_karan.fit_transform(
    cluster_data_karan
)






# ------------------------------------------------------------
# Elbow method
# ------------------------------------------------------------

inertia_values_karan = []

k_values_karan = range(2, 11)

for k in k_values_karan:

    kmeans_temp_karan = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    kmeans_temp_karan.fit(
        cluster_scaled_karan
    )

    inertia_values_karan.append(
        kmeans_temp_karan.inertia_
    )


plt.figure(figsize=(9,5))

plt.plot(
    list(k_values_karan),
    inertia_values_karan,
    marker="o"
)

plt.title("Elbow Method for K-Means")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Within Cluster Sum of Squares")
plt.xticks(list(k_values_karan))
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Silhouette analysis
# ------------------------------------------------------------

silhouette_values_karan = []

for k in k_values_karan:

    kmeans_temp_karan = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels_temp_karan = kmeans_temp_karan.fit_predict(
        cluster_scaled_karan
    )

    silhouette_values_karan.append(
        silhouette_score(
            cluster_scaled_karan,
            labels_temp_karan
        )
    )


plt.figure(figsize=(9,5))

plt.plot(
    list(k_values_karan),
    silhouette_values_karan,
    marker="o"
)

plt.title("Silhouette Score for K-Means")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(list(k_values_karan))
plt.tight_layout()
plt.show()


# selecting K using highest silhouette score

best_k_karan = list(k_values_karan)[
    np.argmax(silhouette_values_karan)
]




# ------------------------------------------------------------
# Final K-Means model
# ------------------------------------------------------------

kmeans_karan = KMeans(
    n_clusters=best_k_karan,
    random_state=42,
    n_init=20
)

kmeans_labels_karan = kmeans_karan.fit_predict(
    cluster_scaled_karan
)

sample_50_karan["kmeans_cluster"] = (
    kmeans_labels_karan + 1
)





# ------------------------------------------------------------
# K-Means cluster centers
# ------------------------------------------------------------

cluster_centers_karan = pd.DataFrame(
    cluster_scaler_karan.inverse_transform(
        kmeans_karan.cluster_centers_
    ),
    columns=clustering_features_karan
)

cluster_centers_karan.index = [
    "Cluster " + str(i + 1)
    for i in range(best_k_karan)
]




# ------------------------------------------------------------
# PCA visualization of K-Means
# ------------------------------------------------------------

kmeans_pca_karan = PCA(
    n_components=2
)

kmeans_pca_result_karan = kmeans_pca_karan.fit_transform(
    cluster_scaled_karan
)

plt.figure(figsize=(9,6))

sns.scatterplot(
    x=kmeans_pca_result_karan[:, 0],
    y=kmeans_pca_result_karan[:, 1],
    hue=sample_50_karan["kmeans_cluster"],
    palette="tab10",
    s=80
)

plt.title("K-Means Clusters - PCA Projection")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 3-D K-Means visualization
# ------------------------------------------------------------

kmeans_pca_3d_karan = PCA(
    n_components=3
)

kmeans_pca_3d_result_karan = kmeans_pca_3d_karan.fit_transform(
    cluster_scaled_karan
)

fig = plt.figure(figsize=(10,7))

ax = fig.add_subplot(111, projection="3d")

scatter_kmeans_3d_karan = ax.scatter(
    kmeans_pca_3d_result_karan[:, 0],
    kmeans_pca_3d_result_karan[:, 1],
    kmeans_pca_3d_result_karan[:, 2],
    c=kmeans_labels_karan,
    cmap="tab10",
    s=70
)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

plt.title("K-Means Clusters - 3-D PCA Projection")

plt.show()


# ------------------------------------------------------------
# K-Means cluster size
# ------------------------------------------------------------

cluster_sizes_karan = (
    sample_50_karan["kmeans_cluster"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(8,5))

cluster_sizes_karan.plot(
    kind="bar"
)

plt.title("K-Means Cluster Sizes")
plt.xlabel("Cluster")
plt.ylabel("Number of Instances")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# K-Means outlier detection using distance from centroid
# ------------------------------------------------------------

distances_from_centroid_karan = np.min(
    kmeans_karan.transform(cluster_scaled_karan),
    axis=1
)

sample_50_karan["centroid_distance"] = (
    distances_from_centroid_karan
)

kmeans_outliers_karan = (
    sample_50_karan
    .sort_values(
        "centroid_distance",
        ascending=False
    )
    .head(5)
)








# ------------------------------------------------------------
# Distance matrix using Euclidean distance
# ------------------------------------------------------------

euclidean_distance_karan = squareform(
    pdist(
        cluster_scaled_karan,
        metric="euclidean"
    )
)




# ------------------------------------------------------------
# Single linkage
# ------------------------------------------------------------

single_linkage_karan = linkage(
    cluster_scaled_karan,
    method="single",
    metric="euclidean"
)

plt.figure(figsize=(12,6))

dendrogram(
    single_linkage_karan,
    labels=sample_50_karan.index.astype(str),
    leaf_rotation=90
)

plt.title("Hierarchical Clustering - Single Linkage")
plt.xlabel("Instances")
plt.ylabel("Euclidean Distance")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Complete linkage
# ------------------------------------------------------------

complete_linkage_karan = linkage(
    cluster_scaled_karan,
    method="complete",
    metric="euclidean"
)

plt.figure(figsize=(12,6))

dendrogram(
    complete_linkage_karan,
    labels=sample_50_karan.index.astype(str),
    leaf_rotation=90
)

plt.title("Hierarchical Clustering - Complete Linkage")
plt.xlabel("Instances")
plt.ylabel("Euclidean Distance")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Average linkage
# ------------------------------------------------------------

average_linkage_karan = linkage(
    cluster_scaled_karan,
    method="average",
    metric="euclidean"
)

plt.figure(figsize=(12,6))

dendrogram(
    average_linkage_karan,
    labels=sample_50_karan.index.astype(str),
    leaf_rotation=90
)

plt.title("Hierarchical Clustering - Average Linkage")
plt.xlabel("Instances")
plt.ylabel("Euclidean Distance")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Centroid linkage
# ------------------------------------------------------------

centroid_linkage_karan = linkage(
    cluster_scaled_karan,
    method="centroid",
    metric="euclidean"
)

plt.figure(figsize=(12,6))

dendrogram(
    centroid_linkage_karan,
    labels=sample_50_karan.index.astype(str),
    leaf_rotation=90
)

plt.title("Hierarchical Clustering - Centroid Linkage")
plt.xlabel("Instances")
plt.ylabel("Euclidean Distance")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Ward linkage
# ------------------------------------------------------------

ward_linkage_karan = linkage(
    cluster_scaled_karan,
    method="ward"
)

plt.figure(figsize=(12,6))

dendrogram(
    ward_linkage_karan,
    labels=sample_50_karan.index.astype(str),
    leaf_rotation=90
)

plt.title("Hierarchical Clustering - Ward Linkage")
plt.xlabel("Instances")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Compare linkage methods
# ------------------------------------------------------------

linkage_methods_karan = {
    "Single": single_linkage_karan,
    "Complete": complete_linkage_karan,
    "Average": average_linkage_karan,
    "Centroid": centroid_linkage_karan,
    "Ward": ward_linkage_karan
}

hierarchical_scores_karan = {}

for method_name_karan, linkage_matrix_karan in linkage_methods_karan.items():

    labels_hierarchical_karan = fcluster(
        linkage_matrix_karan,
        t=best_k_karan,
        criterion="maxclust"
    )

    hierarchical_scores_karan[method_name_karan] = silhouette_score(
        cluster_scaled_karan,
        labels_hierarchical_karan
    )


hierarchical_score_table_karan = pd.DataFrame(
    hierarchical_scores_karan.items(),
    columns=[
        "Linkage Method",
        "Silhouette Score"
    ]
)




# ------------------------------------------------------------
# Ward clustering visualization
# ------------------------------------------------------------

ward_labels_karan = fcluster(
    ward_linkage_karan,
    t=best_k_karan,
    criterion="maxclust"
)

ward_pca_karan = PCA(
    n_components=2
).fit_transform(
    cluster_scaled_karan
)

plt.figure(figsize=(9,6))

sns.scatterplot(
    x=ward_pca_karan[:, 0],
    y=ward_pca_karan[:, 1],
    hue=ward_labels_karan,
    palette="tab10",
    s=80
)

plt.title("Ward Hierarchical Clustering - PCA Projection")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()






# Bisecting K-Means provides a top-down splitting approach

divisive_karan = BisectingKMeans(
    n_clusters=best_k_karan,
    random_state=42
)

divisive_labels_karan = divisive_karan.fit_predict(
    cluster_scaled_karan
)

sample_50_karan["divisive_cluster"] = (
    divisive_labels_karan + 1
)






# ------------------------------------------------------------
# Divisive clustering visualization
# ------------------------------------------------------------

divisive_pca_karan = PCA(
    n_components=2
).fit_transform(
    cluster_scaled_karan
)

plt.figure(figsize=(9,6))

sns.scatterplot(
    x=divisive_pca_karan[:, 0],
    y=divisive_pca_karan[:, 1],
    hue=divisive_labels_karan + 1,
    palette="tab10",
    s=80
)

plt.title("Divisive Clustering - PCA Projection")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()






# Spectral clustering using a nearest-neighbour similarity graph

spectral_karan = SpectralClustering(
    n_clusters=best_k_karan,
    affinity="nearest_neighbors",
    n_neighbors=10,
    random_state=42,
    assign_labels="kmeans"
)

spectral_labels_karan = spectral_karan.fit_predict(
    cluster_scaled_karan
)


sample_50_karan["spectral_cluster"] = (
    spectral_labels_karan + 1
)






# ------------------------------------------------------------
# Spectral clustering visualization
# ------------------------------------------------------------

spectral_pca_karan = PCA(
    n_components=2
).fit_transform(
    cluster_scaled_karan
)

plt.figure(figsize=(9,6))

sns.scatterplot(
    x=spectral_pca_karan[:, 0],
    y=spectral_pca_karan[:, 1],
    hue=spectral_labels_karan + 1,
    palette="tab10",
    s=80
)

plt.title("Spectral Clustering - PCA Projection")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()






# calculating the Euclidean distance matrix

mst_distance_matrix_karan = squareform(
    pdist(
        cluster_scaled_karan,
        metric="euclidean"
    )
)

mst_karan = minimum_spanning_tree(
    mst_distance_matrix_karan
)

mst_dense_karan = mst_karan.toarray()

mst_edges_karan = np.argwhere(
    mst_dense_karan > 0
)




# ------------------------------------------------------------
# MST visualization using PCA coordinates
# ------------------------------------------------------------

mst_pca_karan = PCA(
    n_components=2
).fit_transform(
    cluster_scaled_karan
)

plt.figure(figsize=(10,7))

plt.scatter(
    mst_pca_karan[:, 0],
    mst_pca_karan[:, 1],
    s=60
)

for edge_karan in mst_edges_karan:

    i_karan = edge_karan[0]
    j_karan = edge_karan[1]

    plt.plot(
        [
            mst_pca_karan[i_karan, 0],
            mst_pca_karan[j_karan, 0]
        ],
        [
            mst_pca_karan[i_karan, 1],
            mst_pca_karan[j_karan, 1]
        ],
        linewidth=1
    )

plt.title("Minimum Spanning Tree of 50 Sample Instances")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.tight_layout()
plt.show()






# comparing different numbers of mixture components

em_results_karan = []

for components_karan in range(2, 7):

    gmm_karan = GaussianMixture(
        n_components=components_karan,
        random_state=42
    )

    gmm_karan.fit(
        cluster_scaled_karan
    )

    em_results_karan.append({
        "Components": components_karan,
        "AIC": gmm_karan.aic(cluster_scaled_karan),
        "BIC": gmm_karan.bic(cluster_scaled_karan)
    })


em_results_karan = pd.DataFrame(
    em_results_karan
)




# ------------------------------------------------------------
# AIC and BIC plot
# ------------------------------------------------------------

plt.figure(figsize=(9,5))

plt.plot(
    em_results_karan["Components"],
    em_results_karan["AIC"],
    marker="o",
    label="AIC"
)

plt.plot(
    em_results_karan["Components"],
    em_results_karan["BIC"],
    marker="s",
    label="BIC"
)

plt.title("EM Model Selection using AIC and BIC")
plt.xlabel("Number of Components")
plt.ylabel("Information Criterion")
plt.legend()
plt.tight_layout()
plt.show()


# selecting model using minimum BIC

best_components_karan = int(
    em_results_karan.loc[
        em_results_karan["BIC"].idxmin(),
        "Components"
    ]
)




# ------------------------------------------------------------
# Final Gaussian Mixture Model
# ------------------------------------------------------------

gmm_final_karan = GaussianMixture(
    n_components=best_components_karan,
    random_state=42
)

gmm_labels_karan = gmm_final_karan.fit_predict(
    cluster_scaled_karan
)

sample_50_karan["em_cluster"] = (
    gmm_labels_karan + 1
)


# ------------------------------------------------------------
# EM clustering visualization
# ------------------------------------------------------------

em_pca_karan = PCA(
    n_components=2
).fit_transform(
    cluster_scaled_karan
)

plt.figure(figsize=(9,6))

sns.scatterplot(
    x=em_pca_karan[:, 0],
    y=em_pca_karan[:, 1],
    hue=gmm_labels_karan + 1,
    palette="tab10",
    s=80
)

plt.title("EM Gaussian Mixture Clustering - PCA Projection")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="EM Cluster")
plt.tight_layout()
plt.show()






# Isolation Forest

isolation_forest_karan = IsolationForest(
    contamination=0.10,
    random_state=42
)

outlier_labels_karan = isolation_forest_karan.fit_predict(
    cluster_scaled_karan
)

sample_50_karan["outlier_status"] = np.where(
    outlier_labels_karan == -1,
    "Outlier",
    "Normal"
)





# ------------------------------------------------------------
# Outlier score visualization
# ------------------------------------------------------------

outlier_scores_karan = (
    -isolation_forest_karan.score_samples(
        cluster_scaled_karan
    )
)

sample_50_karan["outlier_score"] = (
    outlier_scores_karan
)


plt.figure(figsize=(9,6))

sns.scatterplot(
    x=pca_2d_karan.transform(
        cluster_scaled_karan
    )[:, 0],
    y=pca_2d_karan.transform(
        cluster_scaled_karan
    )[:, 1],
    hue=sample_50_karan["outlier_status"],
    style=sample_50_karan["outlier_status"],
    s=90
)

plt.title("Outlier Detection using Isolation Forest")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend(title="Status")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Display detected outliers
# ------------------------------------------------------------










# calculating silhouette scores

final_clustering_scores_karan = {
    "K-Means": silhouette_score(
        cluster_scaled_karan,
        kmeans_labels_karan
    ),

    "Divisive": silhouette_score(
        cluster_scaled_karan,
        divisive_labels_karan
    ),

    "Spectral": silhouette_score(
        cluster_scaled_karan,
        spectral_labels_karan
    ),

    "EM": silhouette_score(
        cluster_scaled_karan,
        gmm_labels_karan
    )
}


# add hierarchical scores

for method_name_karan, linkage_matrix_karan in linkage_methods_karan.items():

    labels_final_karan = fcluster(
        linkage_matrix_karan,
        t=best_k_karan,
        criterion="maxclust"
    )

    final_clustering_scores_karan[
        "Hierarchical - " + method_name_karan
    ] = silhouette_score(
        cluster_scaled_karan,
        labels_final_karan
    )


final_clustering_comparison_karan = pd.DataFrame(
    final_clustering_scores_karan.items(),
    columns=[
        "Clustering Method",
        "Silhouette Score"
    ]
)





# ------------------------------------------------------------
# Final comparison plot
# ------------------------------------------------------------

plt.figure(figsize=(12,6))

sns.barplot(
    data=final_clustering_comparison_karan,
    x="Clustering Method",
    y="Silhouette Score"
)

plt.title("Comparison of Clustering Methods")
plt.xlabel("Clustering Method")
plt.ylabel("Silhouette Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# FINAL SAMPLE DATA WITH CLUSTER LABELS
# ============================================================

print("Final 50 Instance Sample with Cluster Labels")

print(
    sample_50_karan[
        [
            "state",
            "year",
            "fatalities",
            "miles",
            "income",
            "age",
            "kmeans_cluster",
            "divisive_cluster",
            "spectral_cluster",
            "em_cluster",
            "outlier_status"
        ]
    ].head(50)
)

print()
```
